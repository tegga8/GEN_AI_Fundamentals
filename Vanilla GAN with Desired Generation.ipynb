```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Dataset
# ============================================================

(x_train, _), (_, _) = tf.keras.datasets.fashion_mnist.load_data()

x_train = x_train.astype("float32")
x_train = (x_train - 127.5) / 127.5
x_train = x_train.reshape(-1, 784)

print("Training Shape:", x_train.shape)


# ============================================================
# Hyperparameters
# ============================================================

latent_dim = 100
batch_size = 128
epochs = 20


# ============================================================
# Generator
# ============================================================

generator = tf.keras.Sequential([
    layers.Input(shape=(latent_dim,)),

    layers.Dense(256),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(512),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(1024),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(784),
    layers.Activation("tanh")
])

generator.summary()


# ============================================================
# Discriminator
# ============================================================

discriminator = tf.keras.Sequential([
    layers.Input(shape=(784,)),

    layers.Dense(1024),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(512),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(256),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(1),
    layers.Activation("sigmoid")
])

discriminator.summary()


# ============================================================
# GAN
# ============================================================

class GAN(tf.keras.Model):

    def __init__(self, discriminator, generator, latent_dim):
        super().__init__()

        self.discriminator = discriminator
        self.generator = generator
        self.latent_dim = latent_dim

        self.d_loss_tracker = tf.keras.metrics.Mean(name="d_loss")
        self.g_loss_tracker = tf.keras.metrics.Mean(name="g_loss")

    @property
    def metrics(self):
        return [
            self.d_loss_tracker,
            self.g_loss_tracker
        ]

    def compile(self, d_optimizer, g_optimizer, loss_fn):

        super().compile()

        self.d_optimizer = d_optimizer
        self.g_optimizer = g_optimizer
        self.loss_fn = loss_fn

    def train_step(self, real_images):

        batch_size = tf.shape(real_images)[0]

        # ====================================================
        # Train Discriminator
        # ====================================================

        random_latent_vectors = tf.random.normal(
            shape=(batch_size, self.latent_dim)
        )

        generated_images = self.generator(
            random_latent_vectors,
            training=True
        )

        combined_images = tf.concat(
            [real_images, generated_images],
            axis=0
        )

        # Real = 1
        # Fake = 0
        real_labels = tf.ones((batch_size, 1))
        fake_labels = tf.zeros((batch_size, 1))

        # One-sided label smoothing
        real_labels = real_labels * 0.9

        labels = tf.concat(
            [real_labels, fake_labels],
            axis=0
        )

        with tf.GradientTape() as tape:

            predictions = self.discriminator(
                combined_images,
                training=True
            )

            d_loss = self.loss_fn(
                labels,
                predictions
            )

        d_grads = tape.gradient(
            d_loss,
            self.discriminator.trainable_weights
        )

        self.d_optimizer.apply_gradients(
            zip(
                d_grads,
                self.discriminator.trainable_weights
            )
        )


        # ====================================================
        # Train Generator
        # ====================================================

        random_latent_vectors = tf.random.normal(
            shape=(batch_size, self.latent_dim)
        )

        # IMPORTANT:
        # Generator wants fake images to be classified as REAL.
        misleading_labels = tf.ones((batch_size, 1))

        with tf.GradientTape() as tape:

            generated_images = self.generator(
                random_latent_vectors,
                training=True
            )

            predictions = self.discriminator(
                generated_images,
                training=True
            )

            g_loss = self.loss_fn(
                misleading_labels,
                predictions
            )

        g_grads = tape.gradient(
            g_loss,
            self.generator.trainable_weights
        )

        self.g_optimizer.apply_gradients(
            zip(
                g_grads,
                self.generator.trainable_weights
            )
        )


        # ====================================================
        # Update Metrics
        # ====================================================

        self.d_loss_tracker.update_state(d_loss)
        self.g_loss_tracker.update_state(g_loss)

        return {
            "d_loss": self.d_loss_tracker.result(),
            "g_loss": self.g_loss_tracker.result()
        }


# ============================================================
# Create GAN
# ============================================================

gan = GAN(
    generator=generator,
    discriminator=discriminator,
    latent_dim=latent_dim
)


gan.compile(
    d_optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0002,
        beta_1=0.5
    ),

    g_optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001,
        beta_1=0.5
    ),

    loss_fn=tf.keras.losses.BinaryCrossentropy()
)


# ============================================================
# Train
# ============================================================

history = gan.fit(
    x_train,
    epochs=epochs,
    batch_size=batch_size
)


# ============================================================
# Plot Loss
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["d_loss"],
    label="Discriminator Loss"
)

plt.plot(
    history.history["g_loss"],
    label="Generator Loss"
)

plt.title("GAN Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)
plt.show()
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. TRAIN A SMALL FASHION-MNIST CLASSIFIER
# ============================================================

(x_cls, y_cls), (_, _) = tf.keras.datasets.fashion_mnist.load_data()

x_cls = x_cls.astype("float32") / 255.0
x_cls = x_cls.reshape(-1, 28, 28, 1)

class_names = [
    "t-shirt",
    "trouser",
    "pullover",
    "dress",
    "coat",
    "sandal",
    "shirt",
    "sneaker",
    "bag",
    "ankle boot"
]

classifier = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),

    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

classifier.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

classifier.fit(
    x_cls,
    y_cls,
    epochs=3,
    batch_size=128,
    validation_split=0.1
)


# ============================================================
# 2. WORD -> FASHION MNIST CLASS
# ============================================================

category_map = {
    "shirt": [6],
    "tshirt": [0],
    "t-shirt": [0],
    "pant": [1],
    "pants": [1],
    "trouser": [1],
    "trousers": [1],
    "pullover": [2],
    "dress": [3],
    "coat": [4],

    # "shoe" means any footwear class
    "shoe": [5, 7, 9],
    "shoes": [5, 7, 9],
    "footwear": [5, 7, 9],

    "sandal": [5],
    "sneaker": [7],
    "bag": [8],
    "ankle boot": [9],
    "boot": [9]
}


# ============================================================
# 3. FIND A GOOD INITIAL LATENT VECTOR
# ============================================================

def find_best_initial_z(target_ids, search_size=512):

    z = tf.random.normal(
        shape=(search_size, latent_dim)
    )

    generated = generator(
        z,
        training=False
    )

    generated_images = (generated + 1.0) / 2.0
    generated_images = tf.reshape(
        generated_images,
        (-1, 28, 28, 1)
    )

    probs = classifier(
        generated_images,
        training=False
    )

    target_score = tf.reduce_sum(
        tf.gather(
            probs,
            target_ids,
            axis=1
        ),
        axis=1
    )

    best_index = tf.argmax(target_score)

    return tf.identity(z[best_index])


# ============================================================
# 4. OPTIMIZE THE LATENT VECTOR
# ============================================================

def find_latent_for_category(
    category,
    steps=400,
    learning_rate=0.05
):

    if category.lower() not in category_map:
        print("Unknown category.")
        print("Available categories:")
        print(list(category_map.keys()))
        return None

    target_ids = category_map[category.lower()]

    # Start from a random latent vector that already looks somewhat
    # like the requested category.
    z = find_best_initial_z(target_ids)

    z = tf.Variable(
        z,
        trainable=True,
        dtype=tf.float32
    )

    optimizer = tf.keras.optimizers.Adam(
        learning_rate=learning_rate
    )

    for step in range(steps):

        with tf.GradientTape() as tape:

            generated = generator(
                tf.expand_dims(z, 0),
                training=False
            )

            generated = (generated + 1.0) / 2.0

            generated = tf.reshape(
                generated,
                (1, 28, 28, 1)
            )

            probs = classifier(
                generated,
                training=False
            )

            target_probability = tf.reduce_sum(
                tf.gather(
                    probs,
                    target_ids,
                    axis=1
                )
            )

            # Encourage z to remain reasonably close to
            # the normal distribution used by the GAN.
            regularization = 0.001 * tf.reduce_mean(
                tf.square(z)
            )

            loss = (
                -tf.math.log(target_probability + 1e-8)
                + regularization
            )

        gradients = tape.gradient(
            loss,
            [z]
        )

        optimizer.apply_gradients(
            zip(gradients, [z])
        )

    return tf.identity(z)
```

```python
def generate(category):

    z = find_latent_for_category(category)

    generated = generator(
        tf.expand_dims(z, 0),
        training=False
    )

    # GAN output: (1, 784)
    generated = (generated + 1.0) / 2.0

    # Convert to image shape for classifier
    generated_for_classifier = tf.reshape(
        generated,
        (1, 28, 28, 1)
    )

    # Display image
    image = generated.numpy()[0].reshape(28, 28)

    # Classify generated image
    probs = classifier(
        generated_for_classifier,
        training=False
    ).numpy()[0]

    predicted_class = np.argmax(probs)

    print("Requested:", category)
    print(
        "Classifier prediction:",
        class_names[predicted_class]
    )
    print(
        "Confidence:",
        round(float(probs[predicted_class]), 4)
    )

    plt.figure(figsize=(4, 4))
    plt.imshow(image, cmap="gray")
    plt.title(
        f"Requested: {category}\n"
        f"Predicted: {class_names[predicted_class]}"
    )
    plt.axis("off")
    plt.show()

    return z, image
```

```python

z_shoe, shoe_image = generate("shoe")
z_shirt, shirt_image = generate("shirt")
z_pant, pant_image = generate("pant")
z_sneaker, sneaker_image = generate("sneaker")
```
